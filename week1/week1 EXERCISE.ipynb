```python
# imports
import os
from dotenv import load_dotenv
from scraper import fetch_website_contents
from IPython.display import Markdown, display, update_display
from openai import OpenAI
```

```python
# set up environment
OPENAI_API_KEY = os.getenv('OPENAI_API_KEY')
OPENAI_BASE_URL = os.getenv('OPENAI_BASE_URL')
OLLAMA_BASE_URL = "http://localhost:11434/v1"

# constants
MODEL_GPT = 'ChatGPT'
MODEL_LLAMA = 'llama3.2'
```

```python
# here is the question; type over this to ask something new

question = """
Explain poisson distribution.
"""

# Get GPT to answer, with streaming
system_prompt = """
You are a helpful assistant that can answer questions about Machine Learning, Data Science, AI, Probability, Statistics, and Python. 
The user has good prior understanding for most of the topics and usually asks questions to clarify the concepts.
You are also a great teacher and can explain complex concepts in a simple way.
You explain in a concise manner and to the point. Also suggest other topics to explore related to the question.
You can also link references for further reading.
"""

user_prompt = f"""
{question}
"""
```

```python
openai = OpenAI(base_url=OPENAI_BASE_URL, api_key=OPENAI_API_KEY)
stream_response = openai.chat.completions.create(
                    model=MODEL_GPT, 
                    messages=[{"role": "system", "content": system_prompt}, 
                    {"role": "user", "content": user_prompt}],
                    stream=True)
# display(Markdown(response.choices[0].message.content))
response = ""
display_handle = display(Markdown(""), display_id=True)
for chunk in stream_response:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)
```

```python
# Get Llama 3.2 to answer

ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')
ollama_stream_response = ollama.chat.completions.create(
                    model=MODEL_LLAMA, 
                    messages=[{"role": "system", "content": system_prompt}, 
                    {"role": "user", "content": user_prompt}],
                    stream=True)
# display(Markdown(ollama_response.choices[0].message.content))
ollama_response = ""
display_handle = display(Markdown(""), display_id=True)
for chunk in ollama_stream_response:
        ollama_response += chunk.choices[0].delta.content or ''
        update_display(Markdown(ollama_response), display_id=display_handle.display_id)
```
